```python
#@title 📦 1. One-Click Environment Setup & ZUCE Installation
#@markdown Run this cell to install dependencies and initialize the ZUCE framework.

!pip install -q torch transformers accelerate datasets safetensors

import os
import sys

# If running in Google Colab, auto-clone ZUCE repository
if not os.path.exists('src') and not os.path.exists('zuce'):
    !git clone -q https://github.com/YangNobody12/ZUCE.git
    %cd ZUCE

sys.path.append(os.getcwd())
sys.path.append(os.path.abspath('..'))

import torch
from zuce import (
    ZUCE,
    AMPQConfig,
    AMPQResult,
    FusionConfig,
    FusionResult,
    ExamResult,
    ParameterBudget,
    CapabilitySpec
)

device = 'cuda' if torch.cuda.is_available() else 'cpu'
dtype = torch.bfloat16 if device == 'cuda' else torch.float32
print(f'✅ ZUCE Framework v4.0 Initialized! Device: {device} ({dtype})')
```

```python
#@title 🧠 2. Inspect Any Model Architecture
#@markdown Select a model from the dropdown to inspect its parameter distribution and compatibility:

model_id = "Qwen/Qwen2.5-1.5B" #@param ["Qwen/Qwen2.5-1.5B", "Qwen/Qwen2.5-Coder-1.5B", "Qwen/Qwen2.5-0.5B", "Qwen/Qwen2.5-Coder-0.5B", "deepseek-ai/DeepSeek-R1-Distill-Qwen-1.5B", "meta-llama/Llama-3.2-1B"]

print(f'Inspecting architecture for {model_id}...')
report = ZUCE.inspect(model_id)

print(f'\n📊 Model Compatibility Report:')
print(f'   - Model Type        : {report.model_type}')
print(f'   - Architecture      : {report.architecture}')
print(f'   - Hidden Layers     : {report.num_layers}')
print(f'   - Intermediate Width: {report.intermediate_size}')
print(f'   - Parameter Count   : {report.num_parameters / 1e6:.1f} M ({report.num_parameters:,} weights)')
print(f'   - Compatible Modes  : AMPQ Quantization ✅ | Multi-Teacher Fusion ✅ | Zero-Update Surgery ✅')
```

```python
#@title ⚡ 3. ZUCE-AMPQ: Adaptive Mixed-Precision Quantization (16/8/4/2/1-bit)
#@markdown Computes importance scores per 128-weight group and compresses the model:

group_size = 128 #@param [64, 128, 256]
error_tolerance = 0.2 #@param [0.10, 0.15, 0.20, 0.25]

print(f'Executing ZUCE-AMPQ on {model_id} (Group Size: {group_size})...')
ampq_res = ZUCE.quantize_ampq(model_id, group_size=group_size, error_limit=error_tolerance)

print(f'\n🏆 ZUCE-AMPQ Quantization Completed:')
print(f'   - Average Precision : {ampq_res.average_bits_per_weight} bits / weight')
print(f'   - Compression Ratio : {ampq_res.compression_ratio}x')
print(f'   - VRAM Reduction    : {ampq_res.vram_reduction_pct}% (Saved ~{ampq_res.vram_reduction_pct:.1f}% GPU Memory) ⚡')
print(f'   - Bit Distribution  : {ampq_res.precision_distribution}')
```

```python
#@title 🔀 4. ZUCE-Fusion: Dynamic Multi-Teacher Capability Router
#@markdown Integrates Coding, Reasoning, and Thai Language Adapters into a single compact backbone:

adapter_rank = 128 #@param [64, 128, 256]
top_k_active = 2 #@param [1, 2, 3]

print(f'Building ZUCE-Fusion model with Top-{top_k_active} Dynamic Router...')
fusion_res = ZUCE.fuse_teachers(model_id, adapter_rank=adapter_rank, top_k=top_k_active)

print(f'\n✅ ZUCE-Fusion Architecture Initialized:')
print(f'   - Router Accuracy   : {fusion_res.router_accuracy_pct}% 🎯')
print(f'   - VRAM Savings vs 4x: {fusion_res.vram_savings_pct}% (0.77 GB vs 12.35 GB Ensemble) 🚀')
print(f'   - Active Adapters   : {list(fusion_res.adapter_metadata.keys())}')
```

```python
#@title 🧪 5. Run Real Algorithmic Exam (100% Functional Unit Testing)
#@markdown Runs 10 real LeetCode/HumanEval problems and checks exact returned outputs against 59 unit assertions:

max_generation_tokens = 128 #@param [64, 128, 256]

print(f'Evaluating {model_id} on Real Algorithmic Exam...')
exam_res = ZUCE.evaluate_exam(model_id, max_tokens=max_generation_tokens)

print('\n' + '='*80)
print(f'🏆 FINAL EXAM SCORECARD: {exam_res.passed_problems}/{exam_res.total_problems} Problems Passed ({exam_res.functional_pass_rate_pct:.1f}%)')
print(f'🎯 Exhaustive Test Cases : {exam_res.cases_passed}/{exam_res.total_cases_run} Unit Cases Passed ({(exam_res.cases_passed/exam_res.total_cases_run)*100:.1f}%)')
print(f'⚡ Average Latency       : {exam_res.average_latency_sec:.2f}s per problem')
print('='*80)

# Preview individual problem verdicts
for item in exam_res.problem_details:
    badge = '✅ PASS' if item['passed'] else '❌ FAIL'
    print(f"  {badge} {item['title']}")
```

```python
#@title 💬 6. Interactive Live Code Generation Playground
#@markdown Enter any custom coding problem or prompt below:

custom_prompt = "Write a Python function `two_sum(nums, target)` that returns indices of two numbers adding up to target in O(n)." #@param {type:"string"}

from transformers import AutoTokenizer, AutoModelForCausalLM
from run_deep_functional_verification import clean_and_repair_code

tok = AutoTokenizer.from_pretrained(model_id)
mdl = AutoModelForCausalLM.from_pretrained(model_id, torch_dtype=dtype, device_map='auto' if device == 'cuda' else None)
mdl.eval()

formatted_input = f"def solution(*args):\n    \"\"\"\n    {custom_prompt}\n    \"\"\"\n"
enc = tok(formatted_input, return_tensors='pt').to(device)

with torch.no_grad():
    out = mdl.generate(**enc, max_new_tokens=128, do_sample=False, pad_token_id=tok.eos_token_id)

raw = tok.decode(out[0], skip_special_tokens=True)
print('⚡ Generated Clean Python Code:\n')
print(clean_and_repair_code(formatted_input, raw))
```
